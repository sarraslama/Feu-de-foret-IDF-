### Évaluation des Seuils MDF (VP, FP, FN)

```python
%matplotlib inline
import matplotlib.pyplot as plt
import numpy as np

plt.rcParams["font.sans-serif"] = "DejaVu Sans"
plt.rcParams["axes.edgecolor"] = "#cbd5e1"
plt.rcParams["axes.linewidth"] = 0.9

# 1. Données d'évaluation des seuils MDF (371 jours couverts en 77)
seuils = ["Seuil niveau_j1 ≥ 2", "Seuil niveau_j1 ≥ 3"]
vp = np.array([6, 0])    # Vrais Positifs
fp = np.array([105, 28])  # Faux Positifs
fn = np.array([25, 31])   # Faux Négatifs

# 2. Affichage des métriques clés dans la console
for i, s in enumerate(seuils):
    precision = vp[i] / (vp[i] + fp[i]) * 100 if (vp[i] + fp[i]) > 0 else 0.0
    rappel = vp[i] / (vp[i] + fn[i]) * 100 if (vp[i] + fn[i]) > 0 else 0.0
    far = fp[i] / (vp[i] + fp[i]) * 100 if (vp[i] + fp[i]) > 0 else 0.0
    print(
        f"{s} -> VP={vp[i]}, FP={fp[i]}, FN={fn[i]} | "
        f"Précision={precision:.1f}% | Rappel={rappel:.1f}% | Taux FP (FAR)={far:.1f}%"
    )

# 3. Tracé du diagramme
fig, ax = plt.subplots(figsize=(10, 6))
fig.suptitle(
    "Projet Feu-de-foret-IDF — Évaluation des Seuils Météo des Forêts (77)",
    fontsize=14.5,
    y=1.01,
    color="#1e293b",
)

x = np.arange(len(seuils))
largeur = 0.25

b1 = ax.bar(x - largeur, vp, width=largeur, label="Vrais Positifs (VP)", color="#16a34a")
b2 = ax.bar(x, fp, width=largeur, label="Faux Positifs (FP)", color="#ea580c")
b3 = ax.bar(x + largeur, fn, width=largeur, label="Faux Négatifs (FN)", color="#dc2626")

for bars in (b1, b2, b3):
    for bar in bars:
        h = int(bar.get_height())
        ax.annotate(
            f"{h}",
            xy=(bar.get_x() + bar.get_width() / 2, h),
            xytext=(0, 4),
            textcoords="offset points",
            ha="center",
            va="bottom",
            fontsize=10,
            color="#111827",
        )

ax.set_title("Faux Positifs & Faux Négatifs par seuil (371 jours couverts en 77)", fontsize=12, pad=12, color="#1e293b")
ax.set_ylabel("Nombre de jours", fontsize=11, color="#334155")
ax.set_xticks(x)
ax.set_xticklabels(seuils, fontsize=11)
ax.set_ylim(0, max(vp.max(), fp.max(), fn.max()) + 12)
ax.grid(axis="y", linestyle="--", alpha=0.55, color="#cbd5e1")
ax.set_axisbelow(True)
ax.tick_params(axis="y", labelsize=11)
ax.legend(frameon=True, facecolor="white", edgecolor="#cbd5e1", fontsize=10, loc="upper right")

# Option d'export haute résolution :
# fig.savefig("evaluation_seuils_mdf.png", dpi=300, bbox_inches="tight")

plt.tight_layout()
plt.show()
```

### Technique Obligatoire : Programmation Linéaire sous Contrainte (MILP)

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt

try:
    from scipy.optimize import milp, LinearConstraint, Bounds
    HAS_SCIPY_MILP = True
except ImportError:
    HAS_SCIPY_MILP = False

plt.rcParams["font.sans-serif"] = "DejaVu Sans"
plt.rcParams["axes.edgecolor"] = "#cbd5e1"
plt.rcParams["axes.linewidth"] = 0.9

# 1. Données des 10 communes prioritaires (77)
communes = [
    "Fontainebleau",
    "Nemours",
    "Provins",
    "Meaux",
    "Melun",
    "Montereau-Fault-Yonne",
    "Coulommiers",
    "Moret-Loing-et-Orvanne",
    "La Ferté-sous-Jouarre",
    "Brie-Comte-Robert",
]
risque = np.array([92, 84, 78, 72, 70, 68, 64, 62, 58, 55], dtype=float)
cout = np.array([18, 15, 14, 13, 12, 12, 11, 11, 10, 10], dtype=float)

budget_max = 80.0          # Budget maximal (k€)
max_patrouilles = 6        # Nombre total de patrouilles disponibles
max_par_commune = 2        # Plafond par commune

n = len(communes)

# 2. Résolution MILP (scipy.optimize.milp ou recherche exacte de secours)
if HAS_SCIPY_MILP:
    res = milp(
        c=-risque,
        integrality=np.ones(n),
        bounds=Bounds(lb=np.zeros(n), ub=np.full(n, max_par_commune)),
        constraints=LinearConstraint(
            np.vstack([cout, np.ones(n)]),
            lb=np.array([-np.inf, -np.inf]),
            ub=np.array([budget_max, max_patrouilles]),
        ),
    )
    x_opt = np.round(res.x).astype(int)
    solveur = "scipy.optimize.milp"
else:
    best_score = -1.0
    x_opt = np.zeros(n, dtype=int)

    def explorer(idx, courant, cout_cum, pat_cum, score_cum):
        global best_score, x_opt
        if idx == n:
            if score_cum > best_score:
                best_score = score_cum
                x_opt = courant.copy()
            return
        for u in range(max_par_commune + 1):
            nc = cout_cum + u * cout[idx]
            np_ = pat_cum + u
            if nc <= budget_max and np_ <= max_patrouilles:
                courant[idx] = u
                explorer(idx + 1, courant, nc, np_, score_cum + u * risque[idx])
        courant[idx] = 0

    explorer(0, np.zeros(n, dtype=int), 0.0, 0, 0.0)
    solveur = "recherche exacte (sans scipy)"

print(f"Solveur utilisé : {solveur}")
print(f"Patrouilles déployées : {x_opt.sum()}/{max_patrouilles}")
print(f"Budget engagé : {np.sum(cout * x_opt):.0f} / {budget_max:.0f} k€")
print(f"Score de risque couvert : {np.sum(risque * x_opt):.0f}")

# 3. Diagramme 1x2 cohérent avec les figures précédentes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6.2))
fig.suptitle(
    "Projet Feu-de-foret-IDF — Allocation Optimale des Patrouilles (MILP, Dépt 77)",
    fontsize=15,
    y=1.02,
    color="#1e293b",
)

# Gauche : Patrouilles affectées par commune
y_pos = np.arange(n)
couleurs_barres = ["#16a34a" if v > 0 else "#94a3b8" for v in x_opt]
bars = ax1.barh(y_pos, x_opt, color=couleurs_barres, height=0.62)

ax1.set_yticks(y_pos)
ax1.set_yticklabels(communes, fontsize=10)
ax1.invert_yaxis()
ax1.set_xlabel(f"Patrouilles affectées (max {max_par_commune} / commune)", fontsize=11, color="#334155")
ax1.set_title(
    f"Affectation optimale ({x_opt.sum()} patrouilles, budget {np.sum(cout * x_opt):.0f} k€)",
    fontsize=12.5,
    pad=12,
    color="#1e293b",
)
ax1.set_xlim(0, max_par_commune + 0.6)
ax1.grid(axis="x", linestyle="--", alpha=0.55, color="#cbd5e1")
ax1.set_axisbelow(True)

for bar, r_val, k_val, x_val in zip(bars, risque, cout, x_opt):
    ax1.annotate(
        f"{x_val} pat. (Risque {int(r_val)}, {int(k_val)}k€)",
        xy=(x_val, bar.get_y() + bar.get_height() / 2),
        xytext=(5, 0),
        textcoords="offset points",
        va="center",
        fontsize=9.5,
        color="#111827",
    )

# Droite : Score de risque couvert vs Coût engagé
x_idx = np.arange(n)
w = 0.38
ax2.bar(x_idx - w / 2, risque * x_opt, width=w, label="Score couvert (Risque × x)", color="#2563eb")
ax2.bar(x_idx + w / 2, cout * x_opt, width=w, label="Coût engagé (k€)", color="#f59e0b")

ax2.set_xticks(x_idx)
ax2.set_xticklabels([c[:10] + "." if len(c) > 10 else c for c in communes], rotation=35, ha="right", fontsize=9.5)
ax2.set_ylabel("Points de risque / Budget (k€)", fontsize=11, color="#334155")
ax2.set_title("Impact et coût engagé par commune", fontsize=12.5, pad=12, color="#1e293b")
ax2.grid(axis="y", linestyle="--", alpha=0.55, color="#cbd5e1")
ax2.set_axisbelow(True)
ax2.legend(frameon=True, facecolor="white", edgecolor="#cbd5e1", fontsize=10, loc="upper right")

plt.tight_layout()
plt.show()
```
