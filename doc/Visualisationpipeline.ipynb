```python
import os
import csv
import glob
import numpy as np
import matplotlib.pyplot as plt

try:
    from scipy.optimize import milp, LinearConstraint, Bounds
    HAS_SCIPY_MILP = True
except ImportError:
    HAS_SCIPY_MILP = False


def charger_csv_simple(chemin):
    """Charge un CSV en liste de dictionnaires sans pandas."""
    with open(chemin, mode="r", encoding="utf-8", errors="replace") as f:
        echantillon = f.read(2048)
        f.seek(0)
        delimiteur = ";" if echantillon.count(";") > echantillon.count(",") else ","
        return list(csv.DictReader(f, delimiter=delimiteur))


def resoudre_milp(scores, budget_max=10, plafond_commune=3):
    """Alloue un nombre entier d'unités pour maximiser sum(scores_i * x_i)."""
    n = len(scores)
    if HAS_SCIPY_MILP:
        res = milp(
            c=-np.asarray(scores, dtype=float),
            integrality=np.ones(n),
            bounds=Bounds(lb=np.zeros(n), ub=np.full(n, plafond_commune)),
            constraints=LinearConstraint(
                np.ones((1, n)),
                lb=np.array([-np.inf]),
                ub=np.array([budget_max]),
            ),
        )
        return np.round(res.x).astype(int)

    x_opt = np.zeros(n, dtype=int)
    reste = budget_max
    for idx in np.argsort(-np.asarray(scores), kind="mergesort"):
        attrib = min(plafond_commune, reste)
        x_opt[idx] = attrib
        reste -= attrib
    return x_opt


def executer_pipeline(dossier_data="data", dossier_sortie="outputs", sauvegarder=False):
    # 1. Vérification éventuelle de fichiers CSV dans data/
    fichiers_csv = glob.glob(os.path.join(dossier_data, "*.csv"))
    nb_lignes_csv = sum(len(charger_csv_simple(p)) for p in fichiers_csv)

    # 2. Données Top 15 communes (77) & paramètres MILP
    communes = [
        "Fontainebleau (77186)", "Recloses (77386)", "Nemours (77333)",
        "Le Vaudoué (77485)", "Noisy-sur-École (77339)", "Donnemarie-Dontilly (77159)",
        "Larchant (77244)", "Arbonne-la-Forêt (77006)", "Bourron-Marlotte (77048)",
        "Vendrest (77490)", "Cerneux (77066)", "La Chapelle-la-Reine (77088)",
        "Forges (77194)", "Évry-Grégy-sur-Yerre (77175)", "Nanteau-sur-Lunain (77329)",
    ]
    scores = np.array([230, 22, 22, 11, 11, 11, 8, 8, 7, 6, 5, 5, 5, 5, 5], dtype=float)

    budget_max = 10
    plafond_commune = 3
    x_opt = resoudre_milp(scores, budget_max=budget_max, plafond_commune=plafond_commune)

    # 3. Données d'évaluation des seuils MDF (371 jours couverts en 77)
    seuils = ["Seuil niveau_j1 ≥ 2", "Seuil niveau_j1 ≥ 3"]
    vp = np.array([6, 0])
    fp = np.array([105, 28])
    fn = np.array([25, 31])

    # 4. Figure récapitulative (Seuils MDF + Allocation MILP)
    plt.rcParams["font.sans-serif"] = "DejaVu Sans"
    plt.rcParams["axes.edgecolor"] = "#cbd5e1"
    plt.rcParams["axes.linewidth"] = 0.9

    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6.2))
    fig.suptitle(
        "Pipeline Feu-de-foret-IDF — Seuils MDF & Allocation MILP (Dépt 77)",
        fontsize=14.5,
        y=1.02,
        color="#1e293b",
    )

    x_idx = np.arange(len(seuils))
    w = 0.25
    b1 = ax1.bar(x_idx - w, vp, width=w, label="VP", color="#16a34a")
    b2 = ax1.bar(x_idx, fp, width=w, label="FP", color="#ea580c")
    b3 = ax1.bar(x_idx + w, fn, width=w, label="FN", color="#dc2626")
    for bars in (b1, b2, b3):
        for bar in bars:
            h = int(bar.get_height())
            ax1.annotate(
                f"{h}",
                xy=(bar.get_x() + bar.get_width() / 2, h),
                xytext=(0, 4),
                textcoords="offset points",
                ha="center",
                va="bottom",
                fontsize=9.5,
            )
    ax1.set_xticks(x_idx)
    ax1.set_xticklabels(seuils, fontsize=10.5)
    ax1.set_ylabel("Nombre de jours", fontsize=10.5)
    ax1.set_title("Évaluation des seuils MDF (371 jours)", fontsize=12, pad=10)
    ax1.grid(axis="y", linestyle="--", alpha=0.55, color="#cbd5e1")
    ax1.set_axisbelow(True)
    ax1.legend(frameon=True, facecolor="white", edgecolor="#cbd5e1")

    top_n = 8
    y_pos = np.arange(top_n)
    couleurs = ["#dc2626" if u > 0 else "#94a3b8" for u in x_opt[:top_n]]
    ax2.barh(y_pos, x_opt[:top_n], color=couleurs, height=0.62)
    ax2.set_yticks(y_pos)
    ax2.set_yticklabels(communes[:top_n], fontsize=9.5)
    ax2.invert_yaxis()
    ax2.set_xlim(0, plafond_commune + 0.7)
    ax2.set_xlabel("Unités allouées (MILP)", fontsize=10.5)
    ax2.set_title(f"Top 8 — Allocation MILP (B={budget_max}, max={plafond_commune})", fontsize=12, pad=10)
    ax2.grid(axis="x", linestyle="--", alpha=0.55, color="#cbd5e1")
    ax2.set_axisbelow(True)

    plt.tight_layout()

    if sauvegarder:
        os.makedirs(dossier_sortie, exist_ok=True)
        chemin_fig = os.path.join(dossier_sortie, "synthese_pipeline_77.png")
        fig.savefig(chemin_fig, dpi=300, bbox_inches="tight")

    plt.show()

    return {
        "fichiers_csv": len(fichiers_csv),
        "lignes_csv": nb_lignes_csv,
        "unites_allouees": int(x_opt.sum()),
        "score_couvert": float(np.sum(scores * x_opt)),
    }


resume = executer_pipeline(sauvegarder=False)
print("Résumé pipeline :", resume)
```
